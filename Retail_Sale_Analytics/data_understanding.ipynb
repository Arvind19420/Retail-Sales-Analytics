```python
import pandas as pd

df = pd.read_csv("online_retail_II.csv", encoding="latin1")

df_clean = df.copy()

df_clean = df_clean.drop_duplicates()

df_clean = df_clean.dropna(subset=['Description'])

df_clean = df_clean[df_clean['Quantity'] > 0]

df_clean = df_clean[df_clean['Price'] > 0]

df_clean['InvoiceDate'] = pd.to_datetime(
    df_clean['InvoiceDate'],
    format='mixed'
)

df_clean['Revenue'] = df_clean['Quantity'] * df_clean['Price']
```

```python
df_clean['Revenue'].sum()

df_clean['Invoice'].nunique()

df_clean['Customer ID'].nunique()

df_clean['Revenue'].sum() / df_clean['Invoice'].nunique()
```

```python
print("Total Revenue:", df_clean['Revenue'].sum())

print("Total Orders:", df_clean['Invoice'].nunique())

print("Total Customers:", df_clean['Customer ID'].nunique())

print("AOV:", df_clean['Revenue'].sum() / df_clean['Invoice'].nunique())
```

```python
df_clean.groupby('Country')['Revenue'].sum()

df_clean.groupby('Country')['Revenue'].sum() \
.sort_values(ascending=False) \
.head(10)
```

```python
print(
    df_clean.groupby('Country')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
print(
    df_clean.groupby('Country')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
df_clean.groupby('Description')['Revenue'].sum() \
.sort_values(ascending=False) \
.head(10)

df_clean.groupby('Description')['Quantity'].sum() \
.sort_values(ascending=False) \
.head(10)
```

```python
df_clean.groupby('Customer ID')['Revenue'].sum() \
.sort_values(ascending=False) \
.head(10)
```

```python
df_clean['Month'] = df_clean['InvoiceDate'].dt.month

df_clean.groupby('Month')['Revenue'].sum()
```

```python
df_clean['Cost'] = df_clean['Revenue'] * 0.70

df_clean['Profit'] = df_clean['Revenue'] - df_clean['Cost']
```

```python
print("Total Profit:", df_clean['Profit'].sum())

print("Average Profit Per Order:",
      df_clean['Profit'].sum() / df_clean['Invoice'].nunique())

print("Profit Margin:",
      (df_clean['Profit'].sum() / df_clean['Revenue'].sum()) * 100)
```

```python
print(
    df_clean.groupby('Country')['Profit']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
print(
    df_clean.groupby('Description')['Profit']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
print(
    df_clean.groupby('Customer ID')['Profit']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
```
